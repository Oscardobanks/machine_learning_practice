```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([ 0.,  0.52631579,  1.05263158,  1.57894737,  2.10526316,
     2.63157895,  3.15789474,  3.68421053,  4.21052632,  4.73684211,
     5.26315789,  5.78947368,  6.31578947,  6.84210526,  7.36842105,
     7.89473684,  8.42105263,  8.94736842,  9.47368421, 10.        ])

Y = np.array([ 2.60943416,  1.76213705,  7.18511292,  9.40744522,  5.46635068,
     8.6061673 , 13.30831239, 14.26225166, 16.70323979, 16.87285951,
    22.17984858, 23.81874177, 24.23732455, 28.20185083, 28.72449237,
    27.91299402, 32.21118578, 31.39802427, 36.91479534, 36.90014818])

X, Y
```

```python
plt.scatter(X, Y, color="blue")
```

```python
# Formal training loop

learning_rate = 0.02
epochs = 10

# initial gradient and intercept
m = 0.3
c = 2

losses = []

y_hat = m * X + c

for i in range(epochs):
  # Calculate the loss.
  y_hat = m * X + c
  loss = np.mean((Y - y_hat)**2)
  losses.append(loss)

  # Calculate the gradient
  dm = -2 * np.mean(X * (Y - y_hat))
  dc = -2 * np.mean(Y - y_hat)

  # Update the parameters
  m = m - learning_rate * dm
  c = c - learning_rate * dc

  # Print losses, gradients and intercepts for n epochs
  print(f"Epochs:  {i}; loss: {loss}")
  print(f"Gradients: {m}; intercepts: {c}")
```

```python
# Plot scatter plot of original X, Y values
# along with a line graph of the final linear model after training
plt.scatter(X, Y, color="magenta")
plt.plot(X, y_hat, color="blue")
plt.show()
```
